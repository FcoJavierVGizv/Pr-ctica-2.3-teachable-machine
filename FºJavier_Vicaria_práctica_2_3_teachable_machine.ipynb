````python
import tensorflow as tf
from keras.models import load_model
from PIL import Image, ImageOps
import numpy as np

class_names = open("labels.txt", "r").readlines()
data = np.ndarray(shape=(1, 224, 224, 3), dtype=np.float32)
image = Image.open("incidente.png").convert("RGB")
size = (224, 224)
image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
image_array = np.asarray(image)
normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
data[0] = normalized_image_array

prediction = model.predict(data)
index = np.argmax(prediction)
class_name = class_names[index]
confidence_score = prediction[0][index]

# Print prediction and confidence score
print("Class:", class_name[2:], end="")
print("Confidence Score:", confidence_score)
````

````python
# implementa la clasificación con la imagen que pertenece claramente a una categoría

image = Image.open("imagen_clara.jpg").convert("RGB")
size = (224, 224)
image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
image_array = np.asarray(image)
normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
data[0] = normalized_image_array

prediction = model.predict(data)
index = np.argmax(prediction)
class_name = class_names[index].strip()
confidence_score = prediction[0][index]

print(f"Clase predicha: {class_name}")
print(f"Confianza: {confidence_score * 100:.2f}%")
````

````python
# implementa la clasificación con una imagen ambigua, que puede ser de cualquier categoría

image = Image.open("imagen_ambigua.jpg").convert("RGB")
size = (224, 224)
image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
image_array = np.asarray(image)
normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
data[0] = normalized_image_array

prediction = model.predict(data)
index = np.argmax(prediction)
class_name = class_names[index].strip()
confidence_score = prediction[0][index]

print(f"Clase predicha: {class_name}")
print(f"Confianza: {confidence_score * 100:.2f}%")
````

````python
# implementa la clasificación con una imagen sin ninguna categoría del modelo

image = Image.open("imagen_desconocida.jpg").convert("RGB")
size = (224, 224)
image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
image_array = np.asarray(image)
normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
data[0] = normalized_image_array

prediction = model.predict(data)
index = np.argmax(prediction)
class_name = class_names[index].strip()
confidence_score = prediction[0][index]

print(f"Clase predicha: {class_name}")
print(f"Confianza: {confidence_score * 100:.2f}%")
````

**¿Cómo se usa el modelo exportado? ¿Cómo se comporta?**

---



Se carga del modelo y etiquetas: Se cargan los pesos del modelo (keras_model.h5) usando la API de compatibilidad de Keras/TensorFlow y se lee el archivo labels.txt con la correspondencia de clases, se abre la imagen de entrada, se convierte a espacio de color RGB y se redimensiona a la resolución esperada (224 \times 224 píxeles). Posteriormente, se transforma la imagen en un array numérico de NumPy y se normalizan sus valores al rango [-1, 1]. Para finalizar se pasa la matriz preprocesada al método model.predict(), la cual devuelve un vector de probabilidades mediante una función Softmax. Con np.argmax() se extrae el índice del valor más alto, asociándolo a la clase predicha y al nivel de confianza.

Comportamientos:

Imagen clara: El modelo clasifica el objeto correctamente con un porcentaje de confianza alto, ya que las características extraídas coinciden de forma evidente con el patrón memorizado durante el entrenamiento.

Imagen ambigua: La confianza disminuye y la puntuación suele distribuirse entre las dos o más clases probables. El modelo predice la opción con mayor peso marginal, pero con mayor incertidumbre.

Imagen fuera de categoría: Los modelos clasificadores cerrados (como Softmax) no reconocen la clase "desconocida" de forma nativa. Obligatoriamente asignan la imagen a una de las clases entrenadas, ofreciendo a veces un nivel de confianza engañosamente alto al basarse únicamente en similitudes de color, bordes o textura del fondo.
